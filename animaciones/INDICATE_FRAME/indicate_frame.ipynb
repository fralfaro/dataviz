```python
# Paso 5: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    grp = VGroup(Text("Hello"), Text("world"))\
      .scale(2).arrange(RIGHT, buff=1)
    
    bg = Rectangle(
      width=config.frame_width+1,
      height=config.frame_height+1,
    )
    w1_bg = Rectangle().surround(grp[0], buff=0.1, stretch=True)
    w2_bg = Rectangle().surround(grp[1], buff=0.1, stretch=True)
    kwargs = dict(fill_opacity=0.95, stroke_width=2, fill_color=BLACK, stroke_color=WHITE)
    bg1 = Cutout(bg, w1_bg, **kwargs)
    bg2 = Cutout(bg, w2_bg, **kwargs)

    self.add(grp)
    self.play(FadeIn(bg1))
    self.wait()
    self.play(ReplacementTransform(bg1, bg2))
    self.wait()
    self.play(FadeOut(bg2))
    self.wait()
%manim $_RI
```

```python
# Paso 6: código de la escena o experimento.

class IndicateFrame(Cutout):
  def __init__(self,
               mob:Mobject,
               buff=0.2,
               width=config.frame_width+1,
               height=config.frame_height+1,
               fill_opacity=0.93,
               stroke_width=2,
               stroke_color=WHITE,
               fill_color=BLACK,
               **kwargs):
    bg = Rectangle(width=width, height=height)
    mob_r = Rectangle().surround(mob, buff=buff, stretch=True)
    super().__init__(bg,
                     mob_r,
                     fill_opacity=fill_opacity,
                     stroke_width=stroke_width,
                     stroke_color=stroke_color,
                     fill_color=fill_color,
                     **kwargs)
  
  def transform(self, mob:Mobject, buff=0.2, **kwargs):
    return Transform(self, IndicateFrame(mob, buff).match_style(self), **kwargs)
```

```python
# Paso 7: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    grp = Tex("Hello", "world", "and", "youtube")\
      .scale(2.5).arrange_in_grid(2, 2, buff=1)
    bg = IndicateFrame(grp[0])

    self.add(grp)
    self.play(FadeIn(bg))
    self.wait()
    self.play(bg.transform(grp[1]))
    self.wait()
    self.play(bg.transform(grp[2]))
    self.wait()
    self.play(bg.transform(grp[3]))
    self.wait()
    self.play(FadeOut(bg))
    self.wait()
%manim $_RI
```
