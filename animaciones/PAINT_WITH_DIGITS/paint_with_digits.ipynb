```python
# Paso 3: código de la escena o experimento.

# pip install mpmath
from mpmath import mp

def get_pi_digits(n):
  mp.dps = n
  return str(mp.pi)
```

```python
# Paso 20: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    config.assets_dir = "./"
    self.camera.background_color = ManimColor("#040921")
    d = Tex(get_pi_digits(36*22-4))[0]
    d.add(*Tex("...")[0])
    d.arrange_in_grid(rows=22, buff=0.05)
    d.set(height=6)
    d.set_color("#2D12DF")
    pi = MathTex("\\pi", color=YELLOW)[0][0]\
      .fade(0.5)\
      .match_height(d[36*2:-36*2])
    for s in d:
      mob = Intersection(pi, s)
      if len(mob.get_all_points()) > 0:
        s.set_color(YELLOW)
    image = ImageMobject("./image_pi").set(height=6)
    grp = Group(d, image).arrange(RIGHT)
    self.add(grp)
%manim $_RI
```
