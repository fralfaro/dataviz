```python
# Paso 11: código de la escena o experimento.

class Cell(VGroup):
  def __init__(self, tex, size=0.7, fade=0.5, height_buff=0.15):
    s = Square(size).fade(fade)
    t = MathTex(tex)\
      .set(height=s.height-height_buff).move_to(s)
    super().__init__(s, t)
```

```python
# Paso 12: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    ROWS = 9
    pyramid = VGroup(*[
      VGroup(*[
        Cell(comb(row, col), height_buff=0.33) # <---
        for col in range(row+1)
      ]).arrange(RIGHT, buff=0)
      for row in range(ROWS)
    ]).arrange(DOWN, buff=0)

    self.add(pyramid)

%manim $_RI
```

```python
# Paso 20: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    ROWS = 7
    pyramid = VGroup(*[
      VGroup(*[
        Cell(str(comb(row, col)), height_buff=0.33, fade=1)
        for col in range(row+1)
      ]).arrange(RIGHT, buff=0)
      for row in range(ROWS)
    ]).arrange(DOWN, buff=0)
    extremes = VGroup(*[
      VGroup(row[0], row[-1]) if len(row) > 1 else row
      for row in pyramid
    ])
    hig = SurroundingRectangle(pyramid[1], buff=0)

    self.play(Write(extremes))
    self.play(Create(hig))
    self.play(
      FadeTransform(VGroup(pyramid[1], hig).copy(), pyramid[2][1])
    )
    for row_up, row_down in zip(pyramid[2:-1], pyramid[3:]):
      for i in range(len(row_up)-1):
        act_row = row_up[i:i+2]
        self.play(
          Transform(hig, SurroundingRectangle(act_row, buff=0))
        )
        self.play(
          FadeTransform(VGroup(act_row, hig).copy(), row_down[i+1])
        )
    self.play(Uncreate(hig))
    self.wait()
%manim $_RV
```
