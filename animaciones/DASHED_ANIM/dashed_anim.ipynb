## 3. Escenas y conceptos

Las siguientes celdas contienen la configuración, las definiciones de escena y los renders interactivos.

```python
# Paso 1: código de la escena o experimento.

from manim import *
config.media_embed = True
config.media_width = "100%"
_RI = "-v WARNING -qm --progress_bar None --disable_caching Example"
```

```python
# Paso 2: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    r = Rectangle(WHITE, 6, 9).round_corners(0.3)
    self.add(r)
%manim $_RI
```

```python
# Paso 3: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    r = Rectangle(WHITE, 6, 9).round_corners(0.3)
    r_dashed = DashedVMobject(r)
    self.add(r_dashed)
%manim $_RI
```

```python
# Paso 4: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    r = Rectangle(WHITE, 6, 9).round_corners(0.3)
    r_dashed = DashedVMobject(r, num_dashes=100)
    self.add(r_dashed)
%manim $_RI
```

```python
# Paso 5: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    r = Rectangle(WHITE, 6, 9).round_corners(0.3)
    r_dashed = DashedVMobject(r, num_dashes=100)
    self.add(r_dashed)

    def updater(mob:Rectangle, alpha:float):
      mob.become(
        DashedVMobject(r, num_dashes=100, dash_offset=alpha)
      )
    
    self.play(UpdateFromAlphaFunc(r_dashed, updater), run_time=5, rate_func=linear)
    self.wait()

%manim $_RI
```

```python
# Paso 6: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    r = Rectangle(WHITE, 6, 9).round_corners(0.3)
    r_dashed = DashedVMobject(r, num_dashes=100).set_stroke(width=10)
    self.add(r_dashed)

    def updater(mob:Rectangle, alpha:float):
      mob.become(
        DashedVMobject(r, num_dashes=100, dash_offset=alpha*20)
      ).set_stroke(width=10)
    
    self.play(UpdateFromAlphaFunc(r_dashed, updater), run_time=5, rate_func=linear)
    self.wait()

%manim $_RI
```
