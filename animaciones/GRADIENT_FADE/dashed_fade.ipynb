```python
# Paso 6: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    s = Square(5, fill_opacity=1)
    s.set_sheen_direction(UL)
    n_colors = 50

    def updater(mob:Square, alpha:float):
      inverse_alpha = 1 - alpha # 1 at start, 0 at end
      n_blacks      = int(inverse_alpha * n_colors)
      n_remaining   = int(n_colors - n_blacks)
      colors        = [BLACK] * n_blacks + [WHITE] * n_remaining
      mob.set_color(colors)

    self.play(UpdateFromAlphaFunc(s, updater), rate_func=linear, run_time=8)
    self.wait()
%manim $_RI
```

```python
# Paso 8: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    s = Square(5, fill_opacity=1)
    s.set_sheen_direction(UL)
    n_colors = 100
    n_extra  = 15
    n_colors_extra  = n_colors + n_extra * 2

    def updater(mob:Square, alpha:float):
      inverse_alpha = 1 - alpha
      extra   = color_gradient([BLACK, WHITE], n_extra)
      n_black = int(inverse_alpha * n_colors_extra)
      n_remaining = n_colors_extra - n_black
      colors = [BLACK] * n_black + extra + [WHITE] * n_remaining
      mob.set_color(colors[n_extra:-n_extra])

    self.play(UpdateFromAlphaFunc(s, updater), rate_func=linear, run_time=8)
    self.wait()
%manim $_RI
```

```python
# Paso 9: código de la escena o experimento.

class Wipe(Animation):
  def __init__(self,
               mob:VMobject,
               mob_color=RED,
               n_extra=20,
               n_colors=100,
               bg_color=BLACK, 
               rate_func=linear,
               **kwargs):
    self.n_extra   = n_extra
    self.n_colors  = n_colors
    self.bg_color  = bg_color
    self.mob_color = mob_color
    self.n_colors_extra = n_colors + n_extra
    super().__init__(mob, rate_func=rate_func, **kwargs)
  
  def interpolate_mobject(self, alpha):
    real_alpha = self.rate_func(alpha)
    inverse_alpha = 1 - real_alpha
    extra   = color_gradient([self.bg_color, self.mob_color], self.n_extra)
    n_black = int(inverse_alpha * self.n_colors_extra)
    n_remaining = self.n_colors_extra - n_black
    colors = [self.bg_color] * n_black + extra + [self.mob_color] * n_remaining
    self.mobject.set_color(colors[self.n_extra:-self.n_extra])

class Example(Scene):
  def construct(self):
    s = Square(5, fill_opacity=1)
    s.set_sheen_direction(UL)
    self.play(Wipe(s, YELLOW), run_time=3)
    self.wait()
%manim $_RI
```
