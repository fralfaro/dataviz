```python
# Paso 9: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    axes = NumberPlane()
    sine = axes.plot(np.sin, x_range=[-PI, PI])
    sine.scale([1.5, 2.5, 1])

    tl = VMobject()
    def tl_updater(mob:TangentLine, alpha:float):
      mob.become(TangentLine(sine, alpha, length=2)).set_color(RED)
    self.add(sine)
    self.play(UpdateFromAlphaFunc(tl, tl_updater), run_time=5, rate_func=linear)
    self.wait()
%manim $_RI
```

```python
# Paso 16: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    fp = SVGMobject("footprints")\
      .scale(0.5).rotate(-PI/2).set_color(WHITE)
    axes = NumberPlane()
    sine = axes.plot(np.sin, x_range=[-PI, PI])
    sine.scale([1.5, 2.5, 1])
    dx = 0.05
    alphas = [*np.arange(0, 1, dx), 1]
    tl_grp = VGroup(*[
      TangentLine(sine, alpha, length=2)
        .rotate(PI/2)
      for alpha in alphas
    ]).set_color(RED)
    grp = VGroup(*[
      fp.copy()
        .move_to(sine.point_from_proportion(alpha))
        .rotate(tl_grp[i].get_angle() - PI/2)
        [i%2]
      for i,alpha in enumerate(alphas)
    ])

    self.play(
      LaggedStart(*[
        FadeIn(mob, rate_func=there_and_back)
        for mob in grp
      ], lag_ratio=0.2),
      run_time=8
    )
    self.wait()

%manim $_RI
```
