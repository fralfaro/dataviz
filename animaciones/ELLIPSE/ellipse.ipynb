## Maths
### Ellipse parametric equation:

$$\LARGE(x,y) = (a\cos\theta, b\sin\theta) \text{ for } 0\leq\theta\leq 2\pi$$

```python
# Paso 3: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    X_SIZE = 8
    Y_SIZE = X_SIZE
    AXIS_LENGTH = 7
    axes = Axes(
      x_range=[-X_SIZE/2, X_SIZE/2, 1],
      y_range=[-Y_SIZE/2, Y_SIZE/2, 1],
      x_length=AXIS_LENGTH,
      y_length=AXIS_LENGTH,
    )
    # VALUES
    a, b = 3, 2 # a = 3, b = 2
    func = lambda t: np.array([
      a * np.cos(t),
      b * np.sin(t),
      0 # z value
    ])
    ellipse = axes.plot_parametric_curve(
      func, t_range=[0, 2*PI, 0.05] # [start, end, resolution]
    )
    self.add(axes, ellipse)

%manim $__
```

## Ellipse focus
![](https://raw.githubusercontent.com/fralfaro/dataviz/main/animaciones/ELLIPSE/ellipse.png)

$$\color{red}\LARGE a^2 = c^2 + b^2 \Rightarrow c = \sqrt{a^2 - b^2}$$

```python
# Paso 5: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    X_SIZE = 8
    Y_SIZE = X_SIZE
    AXIS_LENGTH = 7
    axes = Axes(
      x_range=[-X_SIZE/2, X_SIZE/2, 1],
      y_range=[-Y_SIZE/2, Y_SIZE/2, 1],
      x_length=AXIS_LENGTH,
      y_length=AXIS_LENGTH,
    )
    # VALUES
    a, b = 3, 2 # a = 3, b = 2
    func = lambda t: np.array([
      a * np.cos(t),
      b * np.sin(t),
      0 # z value
    ])
    ellipse = axes.plot_parametric_curve(
      func, t_range=[0, 2*PI, 0.05] # [start, end, resolution]
    )
    # >>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>
    c = np.sqrt(a**2 - b**2)
    focus_left  = Dot(axes.c2p(-c, 0))
    focus_right = Dot(axes.c2p(c, 0))
    # <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
    self.add(axes, ellipse, focus_left, focus_right)

%manim $__
```

## Add updater

```python
# Paso 8: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    X_SIZE = 8
    Y_SIZE = X_SIZE
    AXIS_LENGTH = 7
    axes = Axes(
      x_range=[-X_SIZE/2, X_SIZE/2, 1],
      y_range=[-Y_SIZE/2, Y_SIZE/2, 1],
      x_length=AXIS_LENGTH,
      y_length=AXIS_LENGTH,
    )
    # VALUES
    a, b = 3, 2 # a = 3, b = 2
    func = lambda t: np.array([
      a * np.cos(t),
      b * np.sin(t),
      0 # z value
    ])
    ellipse = axes.plot_parametric_curve(
      func, t_range=[0, 2*PI, 0.05] # [start, end, resolution]
    )
    c = np.sqrt(a**2 - b**2)
    focus_left  = Dot(axes.c2p(-c, 0), color=BLUE)
    focus_right = Dot(axes.c2p(c, 0), color=RED)
    def get_lines(theta): # 0 <= theta <= 2PI
      x, y, _ = func(theta)
      p = axes.c2p(x, y)
      f1p = Line(focus_right.get_center(), p, color=RED) 
      f2p = Line(focus_left.get_center(), p, color=BLUE)
      return [f1p, f2p]
    # >>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>
    f1p, f2p  = get_lines(0)
    lines_grp = VGroup(f1p, f2p)
    def lines_updaters(grp:VGroup, alpha):
      theta = alpha * 2 * PI # or
      # theta = interpolate(0, 2*PI, alpha)
      grp.become(VGroup(*get_lines(theta)))
    # <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
    self.add(
      axes, ellipse, focus_left, focus_right,
      f1p, f2p
    )
    self.play(
      UpdateFromAlphaFunc(lines_grp, lines_updaters),
      run_time=6,
      rate_func=linear
    )
    self.wait()

%manim $__
```

## Using `ValueTracker`

```python
# Paso 10: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    X_SIZE = 8
    Y_SIZE = X_SIZE
    AXIS_LENGTH = 7
    axes = Axes(
      x_range=[-X_SIZE/2, X_SIZE/2, 1],
      y_range=[-Y_SIZE/2, Y_SIZE/2, 1],
      x_length=AXIS_LENGTH,
      y_length=AXIS_LENGTH,
    )
    # VALUES
    a, b = 3, 2 # a = 3, b = 2
    func = lambda t: np.array([
      a * np.cos(t),
      b * np.sin(t),
      0 # z value
    ])
    ellipse = axes.plot_parametric_curve(
      func, t_range=[0, 2*PI, 0.05] # [start, end, resolution]
    )
    c = np.sqrt(a**2 - b**2)
    focus_left  = Dot(axes.c2p(-c, 0), color=BLUE)
    focus_right = Dot(axes.c2p(c, 0), color=RED)
    def get_lines(theta): # 0 <= theta <= 2PI
      x, y, _ = func(theta)
      p = axes.c2p(x, y)
      f1p = Line(focus_right.get_center(), p, color=RED) 
      f2p = Line(focus_left.get_center(), p, color=BLUE)
      return [f1p, f2p]
    # >>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>
    lines_grp = VGroup(*get_lines(0))
    theta_tr  = ValueTracker(0)
    lines_grp.add_updater(
      lambda grp: grp.become(
        VGroup(
          *get_lines(theta_tr.get_value())
        )
      )
    )
    # <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
    self.add(
      axes, ellipse, focus_left, focus_right,
      lines_grp
    )
    self.play(
      theta_tr.animate.set_value(2 * TAU), # TAU = 2 * PI
      run_time=10,
      rate_func=linear
    )
    self.wait()

%manim $__
```

```python
# Paso 13: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    X_SIZE = 8
    Y_SIZE = X_SIZE
    AXIS_LENGTH = 7
    axes = Axes(
      x_range=[-X_SIZE/2, X_SIZE/2, 1],
      y_range=[-Y_SIZE/2, Y_SIZE/2, 1],
      x_length=AXIS_LENGTH,
      y_length=AXIS_LENGTH,
    )
    axes.to_edge(LEFT)
    axes.fade(1)
    # VALUES
    a, b = 3, 2 # a = 3, b = 2
    func = lambda t: np.array([
      a * np.cos(t),
      b * np.sin(t),
      0 # z value
    ])
    ellipse = axes.plot_parametric_curve(
      func, t_range=[0, 2*PI, 0.05] # [start, end, resolution]
    )
    c = np.sqrt(a**2 - b**2)
    focus_left  = Dot(axes.c2p(-c, 0), color=BLUE)
    focus_right = Dot(axes.c2p(c, 0), color=RED)
    def get_lines(theta): # 0 <= theta <= 2PI
      x, y, _ = func(theta)
      p = axes.c2p(x, y)
      f1p = Line(focus_right.get_center(), p, color=RED) 
      f2p = Line(focus_left.get_center(), p, color=BLUE)
      return [f1p, f2p]
    f1p, f2p  = get_lines(0)
    lines_grp = VGroup(f1p, f2p)
    theta_tr  = ValueTracker(0)
    lines_grp.add_updater(
      lambda grp: grp.become(
        VGroup(
          *get_lines(theta_tr.get_value())
        )
      )
    )
    # >>>>>>>>>>>>>>>>>>>>>>>>>>>>>>>
    left_line  = Line(ORIGIN, RIGHT, stroke_width=6)
    right_line = left_line.copy()
    horizontal_lines = VGroup(left_line, right_line)
    def horizontal_lines_updater(grp:VGroup):
      # left_line, right_line = horizontal_lines
      ll, rl   = grp        # right lines
      f1p, f2p = lines_grp  # ellipse lines (left lines)
      rl.set_length(f1p.get_length()).set_color(RED)
      ll.set_length(f2p.get_length()).set_color(BLUE)
      grp.arrange(RIGHT, buff=0).next_to(ellipse, RIGHT, buff=1)

    horizontal_lines.add_updater(horizontal_lines_updater)
    # <<<<<<<<<<<<<<<<<<<<<<<<<<<<<<<
    self.add(
      axes, ellipse, focus_left, focus_right,
      lines_grp, horizontal_lines
    )
    self.play(
      theta_tr.animate.set_value(2 * TAU), 
      run_time=10,
      rate_func=linear
    )
    self.wait()

%manim $__
```
