```python
# Paso 6: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    def get_control_points(vmob:VMobject):
      return VGroup( # *[] not needed from Manim 0.19.0
        Dot(p).scale(0.5)
        for p in vmob.get_all_points()
      )
    l = Line(LEFT, RIGHT).set_length(13)
    dots_before = get_control_points(l).shift(UP).set_color(RED)
    print(f"Before:\t {len(l.get_all_points())=}")
    # Inserting curves
    NumberPlane.prepare_for_nonlinear_transform(l)
    dots_after = get_control_points(l).shift(DOWN).set_color(BLUE)
    print(f"After:\t {len(l.get_all_points())=}")
    self.add(l, dots_before, dots_after)
%manim $_RI
```

```python
# Paso 7: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    plane = NumberPlane(
      x_range=[-6, 6, 1],
      y_range=[-PI, PI, 1],
      faded_line_ratio=5 # thin lines
    )
    plane.add(
      SurroundingRectangle(plane, color=WHITE, buff=0)
    )
    def complex_map(point):
      z = R3_to_complex(point)
      X, Y = float(np.real(z)), float(np.imag(z))
      w = X + np.cos(Y)*0.3 + 1j * np.sin(Y)
      return complex_to_R3(w)

    plane.prepare_for_nonlinear_transform() # <--- important
    plane.apply_function(complex_map)
    self.add(plane)
%manim $_RI
```

```python
# Paso 9: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    plane1 = NumberPlane(
      x_range=[-6, 6, 1],
      y_range=[-PI, PI, 1],
      faded_line_ratio=5 # thin lines
    )
    plane1.add(
      SurroundingRectangle(plane1, color=WHITE, buff=0)
    )
    plane1.prepare_for_nonlinear_transform()
    plane2 = plane1.copy()
    plane0 = plane1.copy()

    plane1.apply_function(
      lambda p: np.array((
        p[0] + np.cos(p[1]) * 0.3,
        np.sin(p[1]),
        0
      ))
    )
    plane2.apply_complex_function(np.exp)

    self.add(plane0)
    self.play(ReplacementTransform(plane0, plane1), rate_func=linear, run_time=4)
    self.wait()
    self.play(ReplacementTransform(plane1, plane2), run_time=4)
    self.wait()
%manim $_RI
```

```python
# Paso 10: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    plane1 = NumberPlane(
      x_range=[-6, 6, 1],
      y_range=[-PI, PI, 1],
      faded_line_ratio=5 # thin lines
    )
    plane1.add(
      SurroundingRectangle(plane1, color=WHITE, buff=0)
    )
    plane1.prepare_for_nonlinear_transform()
    plane2 = plane1.copy()

    plane2.apply_complex_function(np.exp)

    self.play(
      plane1.animate.apply_function(
        lambda p: np.array((
          p[0] + np.cos(p[1]) * 0.3,
          np.sin(p[1]),
          0
        ))
      ),
      run_time=4,
      rate_func=linear
    )
    self.wait()
    self.play(ReplacementTransform(plane1, plane2), run_time=4)
    self.wait()
%manim $_RI
```
