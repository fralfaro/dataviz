```python
# Paso 4: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    c = Circle(3).flip().to_edge(RIGHT, buff=2).fade(0.5)
    d1 = Dot().to_edge(LEFT, buff=2)
    d2 = Dot(c.point_from_proportion(0))
    tr = Triangle(fill_opacity=1)\
      .rotate(PI/6)\
      .scale([0.07, 0.07, 1])
    grp = VMobject()

    l = Line().add_updater(
      lambda m: m.put_start_and_end_on(
        d1.get_center(), d2.get_center()
      ),
      call_updater=True
    )

    dx_lines = 0.2
    def tr_updater(mob:VMobject):
      angle = l.get_angle()
      uv = l.get_unit_vector()
      uv = uv * dx_lines
      starts = [uv]
      vec = ORIGIN.copy()
      while np.linalg.norm(vec) < l.get_length():
        last_start = starts[-1].copy()
        starts.append(last_start + uv)
        vec += uv
      starts.pop()
      # starts.pop()
      mob.become(
        VGroup(*[
          tr.copy()
            .move_to(l.get_start() + v)
            .rotate(angle)
          for v in starts
        ])
      )
    
    grp.add_updater(tr_updater, call_updater=True)

    self.add(c, d1, d2, l, grp)
    self.play(MoveAlongPath(d2, c), run_time=20, rate_func=linear)
    self.wait()
%manim $_RI
```
