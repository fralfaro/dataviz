## Solution

```python
# Paso 9: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    colors = cycle([RED, BLUE, GREEN, PURPLE])
    # polygon
    p = RegularPolygon(4).scale(3)
    vertices = p.get_vertices()
    n_vertices = len(vertices)

    lines = [
      Line(vertices[i], vertices[(i+1)%n_vertices], color=next(colors))
      for i in range(n_vertices)
    ]

    tmp_lines = lines[1:] # last 3 lines
    all_lines = VGroup(*lines)

    # -----
    PROP = 0.3

    for _ in range(20):
      last_line  = tmp_lines[-1]
      first_line = tmp_lines[0]
      next_line = Line(
        last_line.get_end(),
        first_line.point_from_proportion(PROP)
      )
      # remove first line from tmp
      tmp_lines.pop(0)
      # add next_line to the end
      tmp_lines.append(next_line)

      all_lines.add(next_line.set_color(next(colors)))


    self.add(all_lines)

%manim $_RV
```

```python
# Paso 10: código de la escena o experimento.

class SpiralPolygon(VGroup):
  def __init__(self, p:RegularPolygon, n=20, prop=0.3, **kwargs):
    super().__init__(**kwargs)
    vertices = p.get_vertices()
    n_vertices = len(vertices)
    lines = [
      Line(vertices[i], vertices[(i+1)%n_vertices])
      for i in range(n_vertices)
    ]
    tmp_lines = lines[1:]
    for _ in range(n):
      last_line  = tmp_lines[-1]
      first_line = tmp_lines[0]
      next_line = Line(
        last_line.get_end(),
        first_line.point_from_proportion(prop)
      )
      tmp_lines.pop(0)
      tmp_lines.append(next_line)
      self.add(next_line)


class Example(Scene):
  def construct(self):
    p = RegularPolygon(4).scale(3)
    s = SpiralPolygon(p)
    self.add(p, s)

%manim $_RV
```
