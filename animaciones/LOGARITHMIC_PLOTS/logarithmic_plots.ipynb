## 3. Escenas y conceptos

Las siguientes celdas contienen la configuración, las definiciones de escena y los renders interactivos.

```python
# Paso 1: código de la escena o experimento.

from manim import *
config.media_embed = True
config.media_width = "100%"
_RI = "-v WARNING -qm --progress_bar None --disable_caching Example"
```

```python
# Paso 2: código de la escena o experimento.

# Example from: https://docs.manim.community/en/stable/reference/manim.mobject.graphing.coordinate_systems.Axes.html
class Example(Scene):
  def construct(self):
    ax = Axes(
      x_range=[0, 10, 1],
      y_range=[-2, 6, 1],
      tips=False,
      axis_config={"include_numbers": True},
      y_axis_config={"scaling": LogBase(custom_labels=True)},
    )

    # x_min must be > 0 because log is undefined at 0.
    graph = ax.plot(lambda x: x ** 2, x_range=[0.001, 10, 0.02], use_smoothing=False)
    self.add(ax, graph)
%manim $_RI
```

```python
# Paso 3: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    ax = Axes(
      x_range=[0, 5, 1],
      y_range=[0, 6, 1],
      tips=False,
      axis_config={"include_numbers": True},
      x_axis_config={"scaling": LogBase(custom_labels=True)},
    )
    # Axes.c2p(a, b) === Axes@(a, b)
    self.add(ax, Dot(ax@(10**3, 4)))
%manim $_RI
```

```python
# Paso 4: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    ax = Axes(
      x_range=[0, 5, 1],
      y_range=[0, 6, 1],
      tips=False,
      axis_config={"include_numbers": True},
      x_axis_config={"scaling": LogBase(custom_labels=True)},
    )
    #                                         v 10 ** 5
    graph = ax.plot(np.log10, x_range=[0.001, 5, 1], use_smoothing=False)
    self.add(ax, graph)
%manim $_RI
```

```python
# Paso 5: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    ax = Axes(
      x_range=[0, 5, 1],
      y_range=[0, 6, 1],
      tips=False,
      axis_config={"include_numbers": True},
      x_axis_config={
        "scaling": LogBase(custom_labels=False, base=np.e),
        "decimal_number_config": {"num_decimal_places": 3} # 3 decimal digits
      },
    )
    #                                       v e ** 5
    graph = ax.plot(np.log, x_range=[0.001, 5, 1], use_smoothing=False)
    print(f"{np.e**5=}")
    self.add(ax, graph)
%manim $_RI
```

```python
# Paso 6: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    ax = Axes(
      x_range=[0, 5, 1],
      y_range=[0, 6, 1],
      tips=False,
      axis_config={"include_numbers": True},
      x_axis_config={"scaling": LogBase(custom_labels=True)},
    )
    v_lines = VGroup()
    numbers = VGroup()
    for e in range(1,6): # e: exponent from 1 to 5
      for i in range(2,11): # i: Line from 2 to 10
        x = (10 ** e) * (i / 10)
        # x = (10 ** (e - 1)) * i
        l = DashedLine(ax@(x, 0), ax@(x, 6))
        n = Tex(i).scale(0.3).next_to(l, UP, buff=0.05)
        v_lines.add(l)
        numbers.add(n)
    h_lines = VGroup(*[
      DashedLine(ax@(1, i), ax@(10**5, i))
      for i in range(1, 6)
    ])
    graph = ax.plot(np.log10, x_range=[0.001, 5, 1], use_smoothing=False)
    self.add(ax, graph, v_lines, h_lines, numbers)
%manim $_RI
```

```python
# Paso 7: código de la escena o experimento.

class Example(Scene):
  def construct(self):
    ax = Axes(
      x_range=[0, 5, 1],
      y_range=[0, 6, 1],
      tips=False,
      axis_config={"include_numbers": True},
      x_axis_config={"scaling": LogBase(custom_labels=True)},
    )
    v_lines = VGroup()
    for e in range(1,6):
      for i in range(2,11):
        x = (10 ** (e - 1)) * i
        l = Line(ax@(x, 0), ax@(x, 6))
        v_lines.add(l)
    h_lines = VGroup(
      Line(ax@(1, i), ax@(10**5, i))
      for i in range(1, 6)
    )
    grid = VGroup(h_lines, v_lines)
    grid.set_stroke(width=2).fade(0.5)
    graph = ax.plot(np.log10, x_range=[0.001, 5, 1], use_smoothing=False)
    self.add(ax, grid, graph)
%manim $_RI
```
